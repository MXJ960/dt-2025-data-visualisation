```python
sentence = "racecarlevelfield"

# extract first 3 letters
print("Print first 3 letters:", sentence[0:3]) #start: end
print(sentence[:3])
# Extract letters from 5-9
print(sentence[5:9])
#Extract letters from 5th to end
print(sentence[5:])
#Extract the last character
print(sentence[-1])
#Extract last 4 characters
print(sentence[-4:])
#extract first letter of the substring starting from 5
#sentence[5:]: will extract a substring starting from 5
#[0]: this will takes the first character of that substring
print(sentence[5:][0])
#Extract first 5 letters of the substring starting from 5
print(sentence[5:][:5])
#Extract entire String/Text
print(sentence[:])
#Extract every 2nd character
print(sentence[::2])
```

```python
sentence = "Alice went to the park and saw a beautiful bird singing"

words = sentence.split(" ") #Spliit the sentence to words

print(words)
#Extract first word
print(words[0])
#Extract the first 3 words
print(*words[0:3])
#Extract last 3 words
print(words[-3:])
#Print number of words
print("The number of words :", len(words))

```
